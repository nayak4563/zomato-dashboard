```python
af=df.copy()
```

```python
af.drop(columns=["Best Seller"],inplace=True)
```

```python
af.info()
```

```python
af.duplicated().sum()
```

```python
af['Dining Rating'].fillna(af['Dining Rating'].median(), inplace=True)
af['Delivery Rating'].fillna(af['Delivery Rating'].median(), inplace=True)
af['Dining Votes'].fillna(af['Dining Votes'].median(), inplace=True)
af['Delivery Votes'].fillna(af['Delivery Votes'].median(), inplace=True)
```

```python
top_rating_restaurant=af.groupby("Restaurant Name")["Dining Rating"].sum().nlargest(10)
top_rating_restaurant
```

```python
top_deliveryS_restaurant=af.groupby("Restaurant Name")["Delivery Rating"].mean().nlargest(10)
top_deliveryS_restaurant
```

```python
sns.barplot(x=top_rating_restaurant.index,y=top_rating_restaurant.values)
plt.title("Top 10 Rating Restaurant")
plt.xticks(rotation=90)
plt.xlabel("Restaurant Name")
plt.ylabel("Dining Rating")

plt.show()
```

```python
plt.figure(figsize=(8,8))
plt.pie(labels=top_deliveryS_restaurant.index, x=top_deliveryS_restaurant.values, autopct="%1.1f%%")

plt.title("Top 10 Rating Restaurant")
plt.xlabel("Restaurant Name")
plt.ylabel("Delivery Rating")
plt.tight_layout()
plt.show()
```

```python
top_rating_restaurant = af.groupby("Restaurant Name")["Dining Rating"].sum().nlargest(10)

plt.figure(figsize=(10, 6))
sns.barplot(x=top_rating_restaurant.values, y=top_rating_restaurant.index, palette='magma')
plt.title('Top 10 Rated Restaurants (Horizontal Bar Plot)')
plt.xlabel('Sum of Dining Rating')
plt.ylabel('Restaurant Name')
plt.tight_layout()
plt.show()
```

```python
top_voted_restaurant=af.groupby("Restaurant Name")["Dining Votes"].sum().nlargest(5)
top_voted_restaurant
```

```python
sns.barplot(x=top_voted_restaurant.index,y=top_voted_restaurant.values)
plt.title("Top 5 Voted Restaurant")
plt.xticks(rotation=90)
plt.xlabel("Restaurant Name")
plt.ylabel("Dining Votes")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=af,
    x='Dining Rating',
    y='Delivery Rating'
)

plt.title('Dining Rating vs Delivery Rating')
plt.show()
```

```python
af.columns = af.columns.str.strip()

top_cuisines = af["Cuisine"].value_counts().nlargest(5)

plt.figure(figsize=(10, 6))
sns.barplot(x=top_cuisines.values, y=top_cuisines.index, palette='magma')
plt.title('Top 10 Most Common Cuisines')
plt.xlabel('Number of Restaurants')
plt.ylabel('Cuisine')
plt.tight_layout()
plt.show()
```

```python
high_order_cuisine = af.groupby("Cuisine")["Dining Rating"].sum().nlargest(5)
high_order_cuisine
```

```python
sns.barplot(x=high_order_cuisine.index,y=high_order_cuisine.values)
plt.title("High Order Cuisine")
plt.xticks(rotation=90)
plt.xlabel("Cuisine")
plt.ylabel("Dining Rating")

plt.show()
```

```python

top_cities = af['City'].value_counts().nlargest(10).index
top_cuisines_list = af['Cuisine'].value_counts().nlargest(10).index

filtered_df = af[af['City'].isin(top_cities) & af['Cuisine'].isin(top_cuisines_list)]


pivot_df = filtered_df.pivot_table(index='Cuisine', columns='City', aggfunc='size', fill_value=0)

plt.figure(figsize=(12, 8))
sns.heatmap(pivot_df, annot=True, fmt="d", cmap="YlGnBu", cbar_kws={'label': 'Count of Items/Restaurants'})
plt.title('Cuisine Popularity across Top Cities')
plt.xlabel('City')
plt.ylabel('Cuisine')
plt.tight_layout()
plt.show()
```

```python
number_of_cities=af.groupby("City")["Restaurant Name"].nunique()
number_of_cities
```

```python
sns.barplot(x=number_of_cities.index,y=number_of_cities.values)
plt.title("High Rating Restaurant")
plt.xticks(rotation=90)
plt.xlabel("Cuisine")
plt.ylabel("Dining Rating")

plt.show()
```

```python
popular_items = (df.groupby('Item Name')['Votes']
    .sum()
    .sort_values(ascending=False)
    .head(10))

popular_items.plot(kind='barh', figsize=(10, 6),)

plt.title('Top 10 Popular Food Items')
plt.xlabel('Total Votes')
plt.show()
```

```python
top_place=af.groupby

best_seller = af['Place Name'].value_counts()

best_seller.plot(kind='bar', color='red')

plt.title('Place Name')
plt.ylabel('Number of Items')
plt.tight_layout
plt.show()
```

```python
cuisine_price=(af.groupby('Cuisine')['Prices']
    .mean()
    .sort_values(ascending=False)
    .head(10))

cuisine_price.plot(kind='bar', figsize=(10, 5), color='red')

plt.title('Average Price by Cuisine')
plt.ylabel('Average Price')
plt.show()
```
